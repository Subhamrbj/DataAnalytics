# 1. Dataset Loaded Successfully

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

# 3. Dataset Shape and Structure Examined

```python
print("Dataset Shape:", df.shape)
print("\nDataset Information:")
df.info()
```

# 4. Data Types Examined

```python
print("Data Types of Each Column:")
df.dtypes
```

# 5. Missing Values Identified and Handled

```python
print("Missing Values Before Handling:")
print(df.isnull().sum())

# Fill missing numerical values with their median
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("\nMissing Values After Handling:")
print(df.isnull().sum().sum())
```

# 6. Duplicate Records Identified and Handled

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

# 7. Descriptive/Statistical Analysis Performed

```python
df.describe().T
```

# 8. Outliers Investigated

```python
numeric_columns = df.select_dtypes(include=np.number).columns

outlier_summary = {}

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    outlier_summary[column] = len(outliers)

outlier_summary = pd.Series(outlier_summary).sort_values(ascending=False)

print("Number of outliers by numerical column:")
outlier_summary
```

# 9. Univariate Analysis Completed

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Yield_Tonnes_Ha"], kde=True)

plt.title("Distribution of Agricultural Yield")
plt.xlabel("Yield (Tonnes per Hectare)")
plt.ylabel("Frequency")
plt.show()
```

# 10. Bivariate Analysis Completed

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Rainfall vs Agricultural Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes per Hectare)")
plt.show()
```

# 11. Multivariate Analysis Completed

```python
selected_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Soil_Moisture_pct",
    "Fertilizer_kg_ha",
    "Yield_Tonnes_Ha"
]

sns.pairplot(df[selected_columns], diag_kind="hist")

plt.show()
```

# 12. Correlation Analysis Completed

```python
correlation_matrix = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Agricultural Variables")
plt.show()
```

# 13. Seasonal Comparisons Performed

```python
seasonal_summary = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Profit_INR": "mean",
    "Water_Used_m3": "mean"
}).round(2)

seasonal_summary
```

# 14. Student-Designed Analyses

```python
# Analysis 1: Crop-wise Performance
crop_performance = df.groupby("Crop").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Profit_INR": "mean"
}).round(2)

crop_performance
```

```python
# Analysis 2: Irrigation Method Performance
irrigation_performance = df.groupby("Irrigation_Method").agg({
    "Yield_Tonnes_Ha": "mean",
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean"
}).round(2)

irrigation_performance
```

```python
# Analysis 3: Regional Performance
regional_performance = df.groupby("State").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Profit_INR": "mean"
}).sort_values("Yield_Tonnes_Ha", ascending=False).round(2)

regional_performance
```

# 15. Meaningful Insights

```python
print("Key findings from the analysis:\n")

print("1. Average Yield by Season:")
print(df.groupby("Season")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False))

print("\n2. Average Profit by Season:")
print(df.groupby("Season")["Profit_INR"].mean().sort_values(ascending=False))

print("\n3. Average Production by Season:")
print(df.groupby("Season")["Production_Tonnes"].mean().sort_values(ascending=False))

print("\n4. Average Yield by Crop:")
print(df.groupby("Crop")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False))

print("\n5. Average Profit by Crop:")
print(df.groupby("Crop")["Profit_INR"].mean().sort_values(ascending=False))

print("\n6. Average Yield by Irrigation Method:")
print(df.groupby("Irrigation_Method")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False))

print("\n7. Average Water Efficiency by Irrigation Method:")
print(df.groupby("Irrigation_Method")["Water_Efficiency_t_per_1000m3"].mean().sort_values(ascending=False))

print("\n8. Average Disease/Pest Risk by Season:")
print(df.groupby("Season")["Disease_Pest_Risk_pct"].mean().sort_values(ascending=False))
```
